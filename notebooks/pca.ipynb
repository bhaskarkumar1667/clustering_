```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Mall Customers.csv")
```

```python
df
```

```python
for i in df.columns:
    print (df[i].unique())
```

```python
print(df['Education '].value_counts())
print(df['Marital Status'].value_counts())
```

```python
l1={'Unknown':0,'Uneducated':1,'High School':2,'College':3,'Graduate':4,'Post Graduate':5,'Doctorate':6}
l2={'Unknown':1,'Single':1,'Married':2,'Divorced':3}
```

```python
df.drop(['CustomerID'], axis=1, inplace=True)
```

```python
df.columns=df.columns.str.strip()
```

```python
df["Gender"]=df["Gender"].map({'M':1,'F':0})
df["Marital Status"]=df["Marital Status"].map(l2)
df["Education"]=df["Education"].map(l1)
```

```python
df
```

LETS CHECK THE CORRELATION TO SEE IF THERE IS ANY RELATIONSHIP BETWEEN THE VARIABLES. SO THAT WE CAN MERGE THEM.

```python
sns.heatmap(df.corr(),annot=True)
plt.figure(figsize=(20,14))
plt.show()
```

WE CAN'T PERFORM PCA ON ANY OF THESE COLUMNS AS THE CORRELATION BETWEEN THESE COLUMNS IS VERY LESS WHICH SHOWS NO CORRELATION HENCE DATASET WILL REMAIN THE SAME
