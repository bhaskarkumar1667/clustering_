```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans,DBSCAN,AgglomerativeClustering
```

```python
df=pd.read_csv("Mall_Customers_transformed.csv")
df
```

```python
x=df['Annual Income (k$)']
y=df['Spending Score (1-100)']
```

```python
plt.scatter(x,y)
plt.xlabel("income")
plt.ylabel("score")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
```

```python
x_scaled=sc.fit_transform(x.values.reshape(-1,1))
```

```python
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1,11),wcss)
plt.xlabel("k")
plt.ylabel("wcss")
plt.show()
```

```python
km=KMeans(n_clusters=5,init="k-means++",random_state=42)
km.fit(x_scaled)
```

```python
agg=AgglomerativeClustering(n_clusters=5,metric="euclidean",linkage="ward")
agg.fit(x_scaled)
```

```python
db=DBSCAN(eps=5,min_samples=10)
db.fit(x_scaled)
```
