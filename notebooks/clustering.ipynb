```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans,DBSCAN,AgglomerativeClustering
```

```python
df=pd.read_csv("Mall_Customers_transformed.csv")
df
```

```python
x=df[['Annual Income (k$)','Spending Score (1-100)']]
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
```

```python
x_scaled=sc.fit_transform(x)
```

```python
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1,11),wcss)
plt.xlabel("k")
plt.ylabel("wcss")
plt.show()
```

```python
km=KMeans(n_clusters=5,init="k-means++",random_state=42)
```

```python
clusters=km.fit_predict(x_scaled)
```

```python
df["clusters"]=clusters
```

```python
from sklearn.metrics import silhouette_score
sc=silhouette_score(x_scaled,clusters)
print(sc)
```

The Silhouette Score obtained is 0.554, indicating that the clusters are reasonably compact and well separated. This suggests that the selected value of K = 5 produces good-quality clustering for the dataset.

```python
agg=AgglomerativeClustering(n_clusters=5,metric="euclidean",linkage="ward")
agg.fit(x_scaled)
```

```python
db=DBSCAN(eps=5,min_samples=10)
db.fit(x_scaled)
```
